## Czym jest (a czym nie jest) Google Colab?





Wyobraźcie sobie Google Colab jako darmową (z pewnymi ograniczeniami) usługę wynajmu komputera w chmurze Google (tak zwanej *wirtualnej maszyny*). Środowisko to ma już zainstalowany interpreter Pythona oraz wiele przydatnych bibliotek, takich jak Numpy, Pandas, PyTorch i wiele innych.

> **Ważne:** To nie jest wasz lokalny komputer. To tymczasowa **sesja** uruchomiona w chmurze Google. Po jej zamknięciu lub dłuższym okresie nieaktywności, maszyna wirtualna zostaje usunięta, a wszystkie dane, które nie zostały zapisane na stałe, przepadają.

## Google Colab i system Linux


Maszyna wirtualna działa na systemie Liunx, dzięki temu można korzystać z wielu standardowych komend powłoki (shell commands) znanych z Linuksa, takich jak: kopiowanie plików (`cp`), zmiana nazw (`mv`), tworzenie katalogów (`mkdir`) czy wyświetlanie ich zawartości (`ls`).




Wykonanie komendy Linuksowej jest bardzo proste: wystarczy poprzedzić ją wykrzyknikiem (`!`).



> **Przykład**: Jeśli zobaczycie w kodzie zapis `!ls -l`, oznacza to, że wykonana zostanie komenda Linuksa (w tym przypadku szczegółowe listowanie plików), a nie wyrażenie w języku Python. Wykrzyknik jest w tym kontekście sygnałem dla interpretera, by przekazał polecenie do powłoki systemowej.

```python
!ls
```

## Zarządzanie środowiskiem wykonawczym i zmiennymi






| Akcja | Opis / Efekt |
| :--- | :--- |
| Uruchom ponownie sesję (Restart runtime) | Resetuje stan środowiska. Wszystkie zmienne w pamięci zostają usunięte. Pliki zostają! |
| Wyczyść wszystkie dane wyjściowe (Clear all outputs) | Usuwa wyniki wyświetlone pod komórkami kodu (wykresy, wyniki). Nie usuwa zmiennych ani plików. |
| Przerwij wykonywanie kodu (Interrupt execution) | Natychmiast zatrzymuje obecnie wykonywaną komórkę. Stan środowiska (zmienne) zostaje zachowany. |
| Uruchom tę komórkę i poniższe (Run this cell and below) | Wykonuje wskazaną komórkę i automatycznie uruchamia wszystkie komórki poniżej niej w kolejności. |
| Zmień typ środowiska wykonawczego (Change runtime type) | Zmienia sprzęt (np. na GPU). Tworzy nową maszynę wirtualnej, co usuwa wszystkie dane i stan zmiennych. |
| Odłącz się od środowiska wykonawczego i usuń je (Disconnect and delete runtime) | Zamyka sesję i zwalnia zasoby. Stan, pliki i zmienne zostają utracone. |

## Zarządzanie zasobami: CPU vs. GPU




W Colab możecie wybrać "mózg" waszej maszyny. Znajdziecie to w menu: `Środowisko wykonawcze` -> `Zmień typ środowiska wykonawczego`.


### Procesor - CPU (Central Processing Unit)





* Standardowy procesor, jak w waszych laptopach.
* Przydatny do:
    * Wczytywanae i czyszczenie danych (np. biblioteką Pandas).
    * Większości operacji w Scikit-learn (np. Drzewa decyzyjne, SVM, K-Means).
    * Ogólne zadania programistyczne i logiczne.

### Karta graficzna - GPU (Graphics Processing Unit)



* Karta graficzna (w Colabie zazwyczaj Nvidia T4).
* Przydatny do:
    * Treningu sieci neuronowych (TensorFlow/PyTorch).
    * Operacji na dużych macierzach i tensorach (TensorFlow/PyTorch).

💡 **Pro-Tip: Nie marnuj czasu GPU!**

Częsty błąd: "Włączę GPU, wszystko będzie szybsze". **To nieprawda.**

Jeśli wczytujesz dane przez 20 minut używając Pandas (który działa na CPU), twoje GPU w tym czasie ma 0% użycia. Colab ma limit "czasu użycia GPU". Marnując go na zadania CPU, szybciej wyczerpiesz swój limit i możesz dostać niższy priorytet dostępu.

💡 **Pro-Tip: Sprawdź, jakie GPU dostałeś**

```python
!nvidia-smi
```

To polecenie wyświetli Ci model karty (`Name`), jej użycie (`GPU-Util.`) oraz zajętość pamięci (`Memory-Usage`). Jeśli zamiast tych informacji zobaczysz komunikat `command not found`, oznacza to, że Twoje środowisko wykonawcze działa na CPU, a nie na GPU.
